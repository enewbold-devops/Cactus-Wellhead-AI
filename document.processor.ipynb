# Smoke Test - DocLing - Document Chunker

```python
import os
from py_docling_chunks import PDFProcessor
import json
import pandas as pd
_processor = PDFProcessor()

pdf_path = r"C:\Users\AI Engine\Downloads\TRP48079-1 SPECIAL METALS PART 146814  QTY 120-50.pdf"
document, chunks = _processor.process(pdf_path)
'''for chunk in chunks:
    print(chunk)'''

records = []
for chunk in chunks:
    if isinstance(chunk, str):
        records.append(json.loads(chunk))
    elif isinstance(chunk, dict):
        records.append(chunk)
    elif hasattr(chunk, "model_dump"):
        records.append(chunk.model_dump())
    elif hasattr(chunk, "to_dict"):
        records.append(chunk.to_dict())
    else:
        records.append(vars(chunk))

chunks_df = pd.DataFrame(records)
print(chunks_df.shape)
chunks_df.head()
```
